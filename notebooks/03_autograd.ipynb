## 2. What Is Autograd? — 🔥 Interview Essential

Autograd is PyTorch's automatic differentiation system. It records tracked tensor operations during the forward pass and computes gradients during the backward pass.

```text
Input → Model → Prediction → Loss → backward() → Gradients → Parameter update

Forward pass  → calculate predictions and loss
Backward pass → calculate gradients
```


````python
x = torch.tensor(2.0, requires_grad=True)
y = x ** 2
y.backward()
print("x.grad:", x.grad)  # dy/dx = 2x = 4
````

For `y=x²`, `dy/dx=2x`; at `x=2`, the gradient is `4`. Autograd automates this derivative bookkeeping through larger graphs.

## 4. Computational Graphs — 🔥 Interview Essential

PyTorch builds a **dynamic graph** during execution. It remembers enough about tracked operations to differentiate them later, while allowing ordinary Python control flow.

```text
x ──×3──> a ──+4──> b ──square──> y
```

The graph is normally consumed and its saved intermediates released during backward; it is not a permanent artifact.


````python
x = torch.tensor(2.0, requires_grad=True)
a = x * 3
b = a + 4
y = b ** 2
print("values:", x.item(), a.item(), b.item(), y.item())
print("grad_fn:", x.grad_fn, a.grad_fn, b.grad_fn, y.grad_fn)
````

`x` is a leaf created by the user, so `x.grad_fn` is normally `None`. Results of tracked operations are non-leaves and have a `grad_fn`.

## 9. Gradient Accumulation — 🔥 Interview Essential

PyTorch **adds** new gradients into `.grad`; it does not replace them automatically. This enables deliberate accumulation across mini-batches but requires clearing between independent steps.


````python
x = torch.tensor(2., requires_grad=True)
y = x ** 2
y.backward()
print("after first backward:", x.grad)
y = x ** 2  # build a fresh graph
y.backward()
print("after second backward:", x.grad)  # 4 + 4 = 8
````

Typical order:

```text
optimizer.zero_grad() → forward → loss → loss.backward() → optimizer.step()
```

`optimizer.zero_grad()` clears prior parameter gradients so an independent training step starts clean.

| Tool | Scope | Meaning |
|---|---|---|

## 17. Autograd and Neural Network Parameters — 🔥 Interview Essential

`loss.backward()` computes gradients for every trainable model parameter that contributed to the loss.


````python
torch.manual_seed(42)
model = nn.Linear(2,1)
for name,param in model.named_parameters():
    print(name,param.shape,param.requires_grad)
X = torch.tensor([[1.,2.]])
target = torch.tensor([[1.]])
prediction = model(X)
loss = ((prediction-target)**2).mean()
loss.backward()
for name,param in model.named_parameters():
    print(name,"grad:",param.grad)
````

```text
backward() → calculates gradients
step()     → updates parameters
```

Backward does **not** update parameters by itself.

## 18. Manual Gradient Descent

Parameter updates belong inside `no_grad` so the update itself is not recorded. Then clear gradients before the next independent step.


````python
w = torch.tensor(0., requires_grad=True)
x = torch.tensor(2.)
target = torch.tensor(4.)
prediction = w*x
loss = (prediction-target)**2
loss.backward()
print("gradient:",w.grad)
with torch.no_grad():
    w -= 0.1*w.grad
w.grad.zero_()
print("updated w:",w,"cleared grad:",w.grad)
````

### Tiny manual training loop

The cycle is **forward → loss → clear old gradients → backward → update → repeat**. Here the previous gradient is cleared at the end of every iteration.


````python
X = torch.tensor([1.,2.,3.,4.])
y = torch.tensor([2.,4.,6.,8.])
w = torch.tensor(0.,requires_grad=True)
for epoch in range(100):
    prediction=w*X
    loss=((prediction-y)**2).mean()
    loss.backward()
    with torch.no_grad():
        w -= 0.05*w.grad
    w.grad.zero_()
print("learned w:",w.item(),"final loss:",loss.item())
````

### Reusing a freed graph — 🟡 Medium

Calling `backward()` twice on the exact same graph normally fails because saved intermediates are released after the first call. `retain_graph=True` preserves them, but should not be added merely to silence an unexpected error. Usually build a fresh forward graph for the next step.


## 19. Common Autograd Mistakes

1. Forgetting `requires_grad=True`, then expecting `.grad`.
2. Forgetting `backward()`; a computed loss does not calculate gradients automatically.
3. Forgetting to clear accumulated gradients.
4. Confusing `backward()` (calculate gradients) with `step()` (update parameters).

### 1. What is PyTorch Autograd?

<details><summary>Show answer</summary>

**Short Interview Answer:** PyTorch's automatic differentiation system for computing gradients through tensor operations.

**Detailed Explanation:** It records a dynamic graph during tracked forward computation and applies the chain rule in reverse during backward.

</details>


### 2. What does `requires_grad=True` do?

<details><summary>Show answer</summary>

**Short Interview Answer:** It tells PyTorch to track operations involving that tensor for gradient computation.

**Detailed Explanation:** It is normally enabled for trainable floating-point parameters, while inputs or frozen parameters may not need it.

</details>


### 3. What is a computational graph?

<details><summary>Show answer</summary>

**Short Interview Answer:** A graph connecting tensor operations and dependencies used to compute gradients.

**Detailed Explanation:** Nodes represent tensors/operations and edges represent how values depend on earlier values.

</details>


### 4. What is a dynamic computational graph?

<details><summary>Show answer</summary>

**Short Interview Answer:** A graph built as Python code executes during each forward pass.

**Detailed Explanation:** It naturally supports changing control flow and is normally rebuilt for the next forward pass.

</details>


### 5. What does `.backward()` do?

<details><summary>Show answer</summary>

**Short Interview Answer:** It runs reverse-mode differentiation from an output and accumulates gradients into relevant leaves.

**Detailed Explanation:** For a scalar loss, it seeds the output derivative with one and follows the graph backward.

</details>


### 6. Where are gradients stored?

<details><summary>Show answer</summary>

**Short Interview Answer:** Usually in the `.grad` field of leaf tensors and model parameters requiring gradients.

**Detailed Explanation:** Intermediate non-leaf gradients are not retained by default unless `retain_grad()` is requested.

</details>


### 7. Forward vs backward pass?

<details><summary>Show answer</summary>

**Short Interview Answer:** Forward computes predictions and loss; backward computes derivatives of that loss.

**Detailed Explanation:** Parameter updates are a separate optimizer or manual step after backward.

</details>


### 8. Why does PyTorch accumulate gradients?

<details><summary>Show answer</summary>

**Short Interview Answer:** It adds into `.grad`, enabling contributions from multiple paths or mini-batches.

**Detailed Explanation:** The same behavior means independent steps must clear old gradients.

</details>


### 9. Why call `optimizer.zero_grad()`?

<details><summary>Show answer</summary>

**Short Interview Answer:** Because PyTorch accumulates into `.grad`; zero_grad clears the prior step before the next backward.

**Detailed Explanation:** Omitting it unintentionally mixes gradients across training steps unless accumulation is deliberate.

</details>


### 10. What does `.grad` contain?

<details><summary>Show answer</summary>

**Short Interview Answer:** The accumulated derivative of the latest backward objectives with respect to that tensor.

**Detailed Explanation:** Its shape matches the leaf parameter and it can be `None` before any contributing backward pass.

</details>


### 11. What is a leaf tensor?

<details><summary>Show answer</summary>

**Short Interview Answer:** A tensor generally created directly rather than produced by another tracked operation.

**Detailed Explanation:** Trainable model parameters are leaves, and leaf gradients are retained automatically.

</details>


### 12. Why might a non-leaf have `.grad=None`?

<details><summary>Show answer</summary>

**Short Interview Answer:** PyTorch does not retain intermediate gradients by default.

**Detailed Explanation:** Call `retain_grad()` before backward if an intermediate gradient is genuinely needed for inspection.

</details>


### 13. What does `torch.no_grad()` do?

<details><summary>Show answer</summary>

**Short Interview Answer:** It disables gradient recording for operations inside a block.

**Detailed Explanation:** Outputs produced there have no graph connection for those operations, reducing inference memory and work.

</details>


### 14. When use `torch.no_grad()`?

<details><summary>Show answer</summary>

**Short Interview Answer:** During inference, validation, evaluation, and manual parameter updates.

**Detailed Explanation:** Use it whenever operations should not contribute to a backward graph.

</details>


### 15. What does `.detach()` do?

<details><summary>Show answer</summary>

**Short Interview Answer:** It returns a tensor disconnected from the current computational graph.

**Detailed Explanation:** The value is preserved, but later computations from the detached tensor cannot backpropagate through prior history.

</details>


### 16. `detach()` vs `no_grad()`?

<details><summary>Show answer</summary>

**Short Interview Answer:** Detach disconnects a particular tensor; no_grad disables tracking for all operations in a block.

**Detailed Explanation:** They solve related but differently scoped graph-control needs.

</details>


### 17. What happens if you detach part of a graph?

<details><summary>Show answer</summary>

**Short Interview Answer:** Gradient flow stops at the detached tensor.

**Detailed Explanation:** Earlier parameters on that path receive no gradient contribution from later operations.

</details>


### 18. How do you freeze parameters?

<details><summary>Show answer</summary>

**Short Interview Answer:** Set their `requires_grad` to `False`.

**Detailed Explanation:** Commonly loop over a module's parameters, then optionally re-enable selected layers for fine-tuning.

### 19. Why freeze parameters?

<details><summary>Show answer</summary>

**Short Interview Answer:** To keep pretrained features fixed while training fewer parameters.

**Detailed Explanation:** It reduces gradient computation and supports transfer learning, frozen backbones, and selective fine-tuning.

</details>


### 20. `loss.backward()` vs `optimizer.step()`?

<details><summary>Show answer</summary>

**Short Interview Answer:** Backward calculates gradients; step updates parameters using those gradients.

**Detailed Explanation:** They are separate phases, with zero_grad normally preparing the next independent backward.

### 21. Why is loss usually scalar?

<details><summary>Show answer</summary>

**Short Interview Answer:** A scalar gives backward an unambiguous starting derivative of one.

**Detailed Explanation:** Vector outputs require a reduction or an explicit upstream gradient.

</details>


### 22. What happens if backward is called twice?

<details><summary>Show answer</summary>

**Short Interview Answer:** Gradients accumulate, and reusing the exact same freed graph usually errors.

**Detailed Explanation:** A fresh forward graph supports another pass; retain_graph preserves a graph only when intentionally necessary.

</details>


### 23. What does `retain_graph=True` do?

<details><summary>Show answer</summary>

**Short Interview Answer:** It prevents backward from freeing the graph's saved intermediates.

**Detailed Explanation:** It costs memory and should be used for legitimate repeated differentiation, not as a blanket error fix.

</details>


### 24. Why can in-place operations cause errors?

<details><summary>Show answer</summary>

**Short Interview Answer:** They can overwrite values Autograd saved for backward.

**Detailed Explanation:** Version checks detect many unsafe changes; prefer out-of-place operations in tracked code.

</details>


### 25. How convert a tracked tensor to NumPy?

<details><summary>Show answer</summary>

**Short Interview Answer:** Use `x.detach().cpu().numpy()`.

**Detailed Explanation:** Detach removes graph history, CPU ensures compatible memory location, and numpy performs the conversion.

</details>

### Ten manual gradient checks

Predict first; then run the cell to make Autograd verify all ten.

1. `y=5x`, `x=3`
2. `y=x²+2x`, `x=4`
3. `y=x³`, `x=2`
4. `y=(2x+1)²`, `x=1`
5. `y=exp(x)`, `x=0`
6. `y=log(x)`, `x=2`
7. `z=xy`, `x=2,y=5` (both partials)
8. `z=x²y`, `x=3,y=2` (both partials)
9. `y=sqrt(x)`, `x=4`
10. `y=1/x`, `x=4`


````python
checks=[]
def scalar_grads(fn,*values):
    variables=[torch.tensor(float(v),requires_grad=True) for v in values]
    out=fn(*variables)
    out.backward()
    return tuple(v.grad.item() for v in variables)

checks.append(scalar_grads(lambda x:5*x,3))
checks.append(scalar_grads(lambda x:x**2+2*x,4))
checks.append(scalar_grads(lambda x:x**3,2))
checks.append(scalar_grads(lambda x:(2*x+1)**2,1))
checks.append(scalar_grads(lambda x:torch.exp(x),0))
checks.append(scalar_grads(lambda x:torch.log(x),2))
checks.append(scalar_grads(lambda x,y:x*y,2,5))
checks.append(scalar_grads(lambda x,y:x**2*y,3,2))
checks.append(scalar_grads(lambda x:torch.sqrt(x),4))
checks.append(scalar_grads(lambda x:1/x,4))
print(checks)
````

````python
expected=[(5.,),(10.,),(12.,),(12.,),(1.,),(.5,),(5.,2.),(12.,9.),(.25,),(-.0625,)]
assert len(checks)==len(expected)
for actual,wanted in zip(checks,expected):
    assert all(abs(a-b)<1e-6 for a,b in zip(actual,wanted)), f"Mismatch: {actual} vs {wanted}"
print("✅ All ten manual gradients agree with Autograd!")
````

## 25. Autograd Cheat Sheet

```python
# Enable / inspect tracking
x=torch.tensor(2.,requires_grad=True)
x.requires_grad
x.requires_grad_(True)

# Forward, backward, gradient
y=x**2
y.backward()
x.grad  # equivalently: tensor.grad
x.grad.zero_()

# Disable or cut tracking
with torch.no_grad():
    prediction=model(X)
detached=x.detach()

# Freeze / re-enable
for p in model.parameters(): p.requires_grad=False
for p in model.parameters(): p.requires_grad=True

# Typical training flow
optimizer.zero_grad()
prediction=model(X)
loss=loss_fn(prediction,y)
loss.backward()
optimizer.step()
```

```text
backward() → computes gradients
zero_grad() → clears accumulated gradients
step()     → updates parameters
no_grad()  → temporarily disables tracking
detach()   → disconnects one tensor from the current graph
```


### Important mental model

```text
Forward: parameters → model → prediction → loss
Backward: loss → backward() → parameter.grad
Update: parameter.grad → optimizer.step() → new parameters
```
