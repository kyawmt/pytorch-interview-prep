## 2. What Is a Tensor?

A tensor is a multidimensional array with a shape, data type, and device.

| Rank | Common name | Example shape |
|---:|---|---|
| 0D | scalar | `()` |
| 1D | vector | `(3,)` |
| 2D | matrix | `(2, 2)` |
| 3D+ | higher-dimensional tensor | `(2, 2, 2)` |

> **Interview note:** Deep-learning dimensions often represent batch, features, channels, height, width, sequence length, or embedding dimension.


````python
scalar = torch.tensor(5)
vector = torch.tensor([1, 2, 3])
matrix = torch.tensor([[1, 2], [3, 4]])
tensor_3d = torch.tensor([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

for name, tensor in [("scalar", scalar), ("vector", vector), ("matrix", matrix), ("tensor_3d", tensor_3d)]:
    print(f"{name:9} shape={tuple(tensor.shape)}, ndim={tensor.ndim}")
````

## 7. Basic Tensor Operations

`+`, `-`, `*`, `/`, and `**` are element-wise when shapes are compatible. In particular, `x * y` is **not** matrix multiplication.


````python
x = torch.tensor([1.0, 2.0, 3.0])
y = torch.tensor([4.0, 5.0, 6.0])
print("add:", x + y)
print("subtract:", x - y)
print("multiply:", x * y)
print("divide:", x / y)
print("square:", x ** 2)
````

### Reductions

`sum`, `mean`, `max`, and `min` return reduced values. `argmax` and `argmin` return **indices**.


````python
x = torch.tensor([3.0, 7.0, 2.0, 9.0])
print("sum / mean:", torch.sum(x), torch.mean(x))
print("min / argmin:", torch.min(x), torch.argmin(x))
print("max / argmax:", torch.max(x), torch.argmax(x))
````

### `reshape()` vs `view()`

- `view()` requires a compatible/contiguous memory layout and is common for simple reshaping.
- `reshape()` is more flexible; it may return a view or create a copy.

> **Interview answer:** `reshape()` is generally safer when you do not specifically need `view()` semantics.


````python
x = torch.arange(12)
print(x.view(3, 4))
print(x.reshape(2, 6))
````

### `squeeze()` and `unsqueeze()` — 🔥 Interview Essential

`unsqueeze(dim)` inserts a size-1 dimension. `squeeze()` removes size-1 dimensions; `squeeze(dim)` only targets one axis.


````python
x = torch.randn(3, 4)
print("start:", x.shape)
print("unsqueeze(0):", x.unsqueeze(0).shape)  # (1, 3, 4)
print("unsqueeze(1):", x.unsqueeze(1).shape)  # (3, 1, 4)
y = torch.randn(1, 3, 1, 4)
print("squeeze():", y.squeeze().shape)
print("squeeze(0):", y.squeeze(0).shape)
````

## 12. Common Mistakes

1. **`x * y` vs `x @ y`:** `*` is element-wise; `@` is matrix multiplication.
2. **Invalid reshape:** `torch.arange(12).reshape(5, 3)` fails because 12 elements cannot fill 15 positions.
3. **`cat` vs `stack`:** `cat` extends an existing axis; `stack` inserts a new one.
4. **Wrong `dim`:** `softmax`, `argmax`, and concatenation behavior depends on which axis represents classes or groups.
5. **Image layout:** PyTorch image batches are commonly NCHW, not NHWC.

</details>


### 4. `torch.rand()` vs `torch.randn()`?

<details><summary>Show answer</summary>

**Short interview answer:** `rand` samples uniformly from `[0,1)`; `randn` samples a standard normal distribution.

**Explanation:** Choose based on the distribution required, not merely because both are random.

</details>


### 5. `reshape()` vs `view()`?

<details><summary>Show answer</summary>

**Short interview answer:** `view` needs compatible memory layout; `reshape` is more flexible and may copy when necessary.

**Explanation:** Both preserve the number of elements. `reshape` is safer unless view semantics matter.

</details>


### 6. What does `-1` mean in `reshape()`?

<details><summary>Show answer</summary>

**Short interview answer:** It asks PyTorch to infer that dimension from the element count and other dimensions.

**Explanation:** Only one dimension may be inferred.

</details>


### 7. `squeeze()` vs `unsqueeze()`?

<details><summary>Show answer</summary>

**Short interview answer:** `squeeze` removes size-1 axes; `unsqueeze` inserts one.

**Explanation:** Both are common when adapting shapes for batching or channels.

</details>


### 8. `transpose()` vs `permute()`?

<details><summary>Show answer</summary>

**Short interview answer:** `transpose` swaps two axes; `permute` specifies the complete new axis order.

**Explanation:** For NCHW to NHWC, `permute(0,2,3,1)` is the direct expression.

</details>


### 9. `torch.cat()` vs `torch.stack()`?

<details><summary>Show answer</summary>

**Short interview answer:** `cat` joins on an existing axis; `stack` creates a new axis.

**Explanation:** Stack inputs must have identical shapes; cat inputs may differ only along the concatenation axis.

</details>


### 10. What is broadcasting?

<details><summary>Show answer</summary>

**Short interview answer:** It applies operations to compatible shapes by conceptually expanding size-1 or missing leading dimensions.

**Explanation:** For example, adding shape `(4,)` to `(3,4)` applies one bias vector to every row.

</details>


### 11. What does `torch.argmax()` return?

<details><summary>Show answer</summary>

**Short interview answer:** The index of the maximum value, optionally along a selected dimension.

**Explanation:** It returns indices, unlike `torch.max`, which can provide maximum values.

</details>


### 12. Element-wise vs matrix multiplication?

<details><summary>Show answer</summary>

**Short interview answer:** `x*y` multiplies corresponding elements; `x@y` contracts matching inner dimensions.

**Explanation:** For 2D tensors, `(m,n)@(n,p)` produces `(m,p)`.

</details>


### 13. What is the common PyTorch image format?

<details><summary>Show answer</summary>

**Short interview answer:** NCHW: batch, channels, height, width.

**Explanation:** Some visualization tools use NHWC, so `permute` is frequently needed.

</details>


### 14. What dtype is common for class labels?

<details><summary>Show answer</summary>

**Short interview answer:** `torch.long`, which is `torch.int64`.

**Explanation:** Losses such as `CrossEntropyLoss` commonly expect integer class indices.

</details>


### 15. How do you move a tensor to another device?

<details><summary>Show answer</summary>

**Short interview answer:** Create or select a `torch.device`, then call `x = x.to(device)`.

**Explanation:** Model parameters and input tensors must be on compatible devices.

</details>
