## 5. Understanding `dim` — 🔥 Interview Essential

For shape `(2, 3)`, `dim=0` collapses rows and leaves `(3,)`; `dim=1` collapses columns and leaves `(2,)`.

```text
dim=0 ↓ rows collapse     [[1,2,3], [4,5,6]] → [5,7,9]
dim=1 → columns collapse  [[1,2,3], [4,5,6]] → [6,15]
```

The robust rule: **the specified dimension disappears after a reduction unless `keepdim=True`.**


````python
x = torch.tensor([[1, 2, 3], [4, 5, 6]])
for dim in (0, 1):
    result = torch.sum(x, dim=dim)
    print(f"sum dim={dim}: {result.tolist()}, shape={tuple(result.shape)}")
````

### Normalization using broadcasting — 🔥 Interview Essential

````python
x = torch.randn(4, 3)
mean = x.mean(dim=0, keepdim=True)
std = x.std(dim=0, keepdim=True)
normalized = (x - mean) / std
print("x/mean/std:", x.shape, mean.shape, std.shape)
print("normalized feature means:", normalized.mean(dim=0))
````

## 13. Batch Matrix Multiplication — 🔥 Interview Essential for Deep Learning

`bmm` applies a separate matrix multiplication to every batch and explicitly expects 3D tensors.

```text
(32,10,64) @ (32,64,20) → (32,10,20)
batch 0: (10,64) @ (64,20)
...
batch 31: (10,64) @ (64,20)
```

This pattern appears in attention, sequence models, and other batched neural operations.


````python
A = torch.randn(32, 10, 64)
B = torch.randn(32, 64, 20)
C = torch.bmm(A, B)
print(A.shape, B.shape, "->", C.shape)
print("matmul agrees:", torch.allclose(C, torch.matmul(A, B)))
````

## 18. Common Mistakes

1. `argmax(logits, dim=0)` gives one winner per class; classification usually needs `dim=1` for one class per sample.
2. Dropping a reduced axis can make normalization awkward; `keepdim=True` keeps it broadcastable.
3. `(x>0) and (x<10)` is not element-wise; use `(x>0) & (x<10)`.
4. `(3,4) + (3,)` fails because trailing dimensions 4 and 3 conflict.
5. `sort` returns values and indices; `argsort` returns only indices.
6. `max` gives the value; `argmax` gives its index.
7. Concatenating on the wrong axis may join batches when features were intended, or vice versa.
8. `(32,10,64) bmm (32,20,64)` fails: inner dimensions 64 and 20 do not match; the second input should be `(32,64,p)`.

### 1. What is broadcasting?

**Short Interview Answer:** It applies element-wise operations to compatible shapes without manually copying data.

### 2. What are PyTorch broadcasting rules?

<details><summary>Show answer</summary>

**Short Interview Answer:** Compare shapes from the right; each pair must be equal or one must be 1.

**Detailed Explanation:** Missing leading dimensions are treated as 1. The output takes the maximum compatible size on every axis.

</details>


### 3. What does `dim` mean in a reduction?

<details><summary>Show answer</summary>

**Short Interview Answer:** It is the axis being collapsed.

**Detailed Explanation:** That axis disappears unless keepdim is true; all other axes remain in order.

</details>


### 4. What does `keepdim=True` do?

<details><summary>Show answer</summary>

**Short Interview Answer:** It retains reduced axes with size 1.

**Detailed Explanation:** This preserves rank and makes later broadcasting, especially normalization, easier.

</details>


### 5. `max()` vs `argmax()`?

<details><summary>Show answer</summary>

**Short Interview Answer:** `max` returns the largest value; `argmax` returns its index.

**Detailed Explanation:** With a dimension, `max` can return both values and indices, while `argmax` returns indices only.

</details>


### 6. `sort()` vs `argsort()`?

<details><summary>Show answer</summary>

**Short Interview Answer:** `sort` returns sorted values and original indices; `argsort` returns only indices.

</details>


### 7. What does `topk()` return?

**Detailed Explanation:** Both outputs have that dimension replaced by size k.

### 10. What does `torch.where()` do?

**Detailed Explanation:** All three inputs follow broadcasting rules, and the result has their broadcast shape.

</details>


### 11. `cat()` vs `stack()`?

**Detailed Explanation:** Stack requires all input shapes to match exactly.

</details>


### 12. `chunk()` vs `split()`?

### 13. `mm()` vs `matmul()`?

<details><summary>Show answer</summary>

**Short Interview Answer:** `mm` is strictly 2D; `matmul` supports vectors, matrices, batches, and batch broadcasting.

**Detailed Explanation:** Use the narrower operation when it communicates intent, but matmul is more general.

<details><summary>Show answer</summary>

**Short Interview Answer:** It multiplies corresponding matrices in two 3D batches.

**Detailed Explanation:** Shapes `(b,m,n)` and `(b,n,p)` produce `(b,m,p)` with no batch broadcasting.

# Element-wise / math
